```python
import pandas

result = pandas.read_csv("result/result.csv").sort_values(by=["orderid"]).reset_index(drop=True)
answer = pandas.read_csv("result/answer.csv").sort_values(by=["orderid"]).reset_index(drop=True)

result["compare"] = (result["is_late"] == answer["is_late"])
print(result.loc[result.compare == True].shape[0] / result.shape[0] * 100, "% accuracy")
```
