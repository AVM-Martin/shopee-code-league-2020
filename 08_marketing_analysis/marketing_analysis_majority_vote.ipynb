```python
DATA_DIR = "dataset/"
MODEL_DIR = "model/"
CHECKPOINT_DIR = "epoch_model/"
RESULT_DIR = "result/"

PREDICT_CSV = DATA_DIR + "test.csv"
```

```python
import glob


SUBMISSIONS = [
    filename[7:] for filename in glob.glob(
        RESULT_DIR + "s" + 8 * "[0-9]" + "_" + "5" + 4 * "[0-9]" + ".csv"
    )
]
SUBMISSIONS.sort(key=lambda x: x[10:15])


BEST_SUBMISSION = "s16770919_50425.csv"
```

```python
MODEL_NAME = "majority_" + "-".join([submission[:9] for submission in SUBMISSIONS])
```

# Import Data

```python
import pandas


raw_predict = pandas.read_csv(PREDICT_CSV)
```

```python
from collections import Counter


def get_majority_vote(x):
    vote = Counter([x[submission] for submission in SUBMISSIONS] + 2 * [x[BEST_SUBMISSION]])
    result = vote.most_common(2)
    if len(result) > 1 and result[0][1] == result[1][1]:
        return x[BEST_SUBMISSION]

    return result[0][0]
```

```python
RESULT_INDEX = [
    "row_id",
]

RESULT_DATA = [
    "open_flag",
]
```

# Combine

```python
result = raw_predict.copy(deep=True)

for submission in SUBMISSIONS:
    temp = pandas.read_csv(RESULT_DIR + submission)
    result[submission] = temp[RESULT_DATA]
    
result[RESULT_DATA[0]] = result.apply(get_majority_vote, axis=1)
```

# Save Result

```python
result = result[RESULT_INDEX + RESULT_DATA]

result.to_csv(RESULT_DIR + MODEL_NAME + "_result.csv", index=False)
```
